# Data Pre-processing

Data preprocessing is an essential step in data analytics. It improves the quality of the dataset by handling missing values, removing duplicate records, correcting data types, creating new features, and preparing the data for further analysis.

## Handling Missing Values

Missing values can affect analysis and model performance. Therefore, they must be identified and handled appropriately.

```python
df.isnull().sum()
```

```python
# Fill missing values in Age with the mean
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
# Fill missing value in EmploymentType with the mode
df["EmploymentType"] = df["EmploymentType"].fillna(df["EmploymentType"].mode()[0])
```

```python
# Verify that all missing values have been handled
df.isnull().sum()
```

### Observation

Initially, the dataset contained **8 missing values** in the **Age** column and **7 missing value** in the **EmploymentType** column.

The missing values in the **Age** column were replaced with the **mean age**, while the missing value in the **EmploymentType** column was replaced with the **most frequent employment type (mode)**. After preprocessing, no missing values remained in the dataset.

###GroupBy Analysis

The groupby() function is used to divide the dataset into groups based on a categorical variable and calculate summary statistics for each group. In this analysis, employees are grouped by Department to calculate employee count, average monthly income, and average job satisfaction.

This helps compare different departments based on workforce size, income, and satisfaction.

```python
department_summary = df.groupby('Department').agg(
    EmployeeCount=('EmployeeID', 'count'),
    AverageMonthlyIncome=('MonthlyIncome', 'mean')
)

department_summary
```

###Observation:

##Bivariate Analysis
###Bar/Count Plot: Attrition by Overtime

Overtime is an important workplace factor that may influence employee workload and work-life balance. Attrition is compared with Overtime to understand whether employees who work overtime show different attrition patterns from employees who do not work overtime.

### Line Chart: Average Monthly Income by Years at Company

A line chart is used to examine how the average monthly income changes according to the number of years employees have spent in the company. YearsAtCompany represents employee tenure, while MonthlyIncome represents compensation. This visualization helps identify whether average income tends to increase, decrease, or fluctuate as employees gain more experience within the organization.
Matplotlip is used to create an interactive pie chart.




```python
# Calculate average monthly income for each year at the company

income_by_years = (
    df.groupby('YearsAtCompany')['MonthlyIncome']
      .mean()
      .reset_index()
      .sort_values('YearsAtCompany')
)

# Create line chart

plt.figure(figsize=(10,6))

plt.plot(
    income_by_years['YearsAtCompany'],
    income_by_years['MonthlyIncome'],
    marker='o',
    linewidth=2
)

plt.title('Average Monthly Income by Years at Company')
plt.xlabel('Years at Company')
plt.ylabel('Average Monthly Income')
plt.grid(True)

plt.show()
```

### Observation:

The line chart shows how average monthly income varies with employees' years at the company. The average income changes across different levels of employee tenure, showing the relationship between experience within the organization and compensation. This analysis helps identify whether longer tenure is associated with changes in average monthly income.

###Box Plot: Monthly Income vs Attrition

Monthly Income is compared with Attrition to determine whether salary levels differ between employees who stayed and employees who left. A box plot is useful because it displays the median, spread, and distribution of monthly income for each attrition category.

###Violin Plot: Monthly Income by Department and Attrition
A violin plot is used to visualize the distribution of monthly income across different departments while comparing employees based on their attrition status. It combines information about the distribution and spread of income values, making it useful for understanding differences between employees who stayed and employees who left.
The Department variable is placed on the x-axis, while MonthlyIncome is placed on the y-axis. The Attrition variable is represented using different sections within each violin. ###Heatmap: Correlation Analysis

Correlation analysis is used to identify relationships between numerical variables. A correlation heatmap provides a visual representation of these relationships. Correlation values range from -1 to +1, where positive values indicate a positive relationship, negative values indicate an inverse relationship, and values close to zero indicate a weak relationship.

Seaborn is used to create the correlation heatmap.

```python
numeric_df = df.select_dtypes(include='number')

correlation = numeric_df.corr()

plt.figure(figsize=(14,10))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.xlabel('Variables')
plt.ylabel('Variables')

plt.show()
```

###Observation:

The correlation heatmap provides an overall view of relationships among the numerical variables in the dataset. Strong positive correlations indicate variables that tend to increase together, while negative correlations indicate inverse relationships. Variables with weak correlations show limited linear relationships. The heatmap can be used to identify potentially important variables for further analysis and predictive modeling.
